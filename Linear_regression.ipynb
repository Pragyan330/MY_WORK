```python
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score
```

```python
data = pd.read_csv("Train.csv")
data = data.drop(columns = ["Unnamed: 0","Name"])
```

```python
data = data.dropna()
data = data.reset_index()
data.info()
data["Engine"] = data["Engine"].str.replace(" CC"," ").astype(float)
data["Mileage"] = data["Mileage"].str.replace(" kmpl", " ").astype(float)
data["Power"] = data["Power"].str.replace(" bhp", " ")
```

```python
data = data[data["Power"]!= "null "]
```

```python
data["Transmission"] = data["Transmission"].map({"Manual":0,"Automatic": 1})
data["Fuel_Type"] = data["Fuel_Type"].map({"Petrol": 0, "Diesel": 1})
data = pd.get_dummies(data, columns = ["Owner_Type","Location"],dtype = int)
# data = pd.get_dummies(data, columns = [],dtype = int)
```

```python
data = data.drop(columns = ["index"])
```

```python
data = data.dropna()
data = data.reset_index()
data = data.astype(float)
data
```

```python
means = data.mean()
stds = data.std()
std_data = (data-means)/stds
std_data
```

```python
x_train = std_data.drop(columns = ["Price"])
y_train = std_data["Price"]
```

```python
lr = 0.0001
rows,features = x_train.shape
weight = np.zeros(features)
bias=0
    
for i in range(1000):
    y_pred = np.dot(x_train,weight)+ bias
    db =   np.sum((y_pred - y_train))/(rows)
    dw = np.dot(x_train.T,(y_pred - y_train))
    weight = weight - lr*dw
    bias = bias - lr*db
    cost = np.sum((y_pred - y_train)**2)/(2*rows)
    if i%100 == 0:
        print(cost)
```
