```python
#this randomly initializes the weights and bias for the beginning the whole thing
def weight_bias_make(no_of_neurons,inputs,number):
    return np.random.randn(no_of_neurons,inputs)*0.01, 0
#this is for getting answers before getting into the next layer by dot product
def forward_prop(weights,input_data,bias):
    return np.dot(weights,input_data) + bias

def activation_function(Z):
    #this is the activation function to get the values squished
    return (1/(1+ np.exp(-Z)))
def relu(x):
    return np.maximum(0, x)

def softmax(x):
    # shift input for numerical stability
    x_shifted = x - np.max(x, axis=0, keepdims=True)
    
    # sompute the exponentials
    exp_x = np.exp(x_shifted)
    
    # normalize by the sum of exponentials for each column
    softmax_probs = exp_x / np.sum(exp_x, axis=0, keepdims=True)
    return softmax_probs.T
def ReLU_deriv(Z):
    return Z > 0
```

```python
# this whole cell is dedicated to initialized the weight,bias for layer no. 1 then get the output and sigmoid squishify the output

weight_layer1,bias_layer1 = weight_bias_make(100,784,4000)

unactivated_output1 = forward_prop(weight_layer1,images,bias_layer1)
neuron_output_layer1 = relu(unactivated_output1)
weight_layer1
```

```python
# this whole cell is dedicated to initialized the weight,bias for layer no. 2 then get the output and sigmoid squishify the output
weight_layer2,bias_layer2 = weight_bias_make(10,100,4000)
unactivated_output2 = forward_prop(weight_layer2,neuron_output_layer1,bias_layer2)
neuron_output_layer2 = relu(unactivated_output2)
neuron_output_layer2
```

```python
# this whole cell is dedicated to initialized the weight,bias for layer no. 3 then get the output and sigmoid squishify the output
weight_layer3,bias_layer3 =  weight_bias_make(5,10,4000)
unactivated_output3 = forward_prop(weight_layer3,neuron_output_layer2,bias_layer3)
print(unactivated_output3.shape)
neuron_output_layer3 = softmax(unactivated_output3) #softmax is being used,'''experimental ZZZZ '''
neuron_output_layer3
```

```python
def back_prop(images,classes,pred3,pred2,pred1,rows,W1,W2,W3,unac_1,unac_2,unac_3):
    #lots of math wrote it while looking at the formulas on the internet, to understand this, have to read
    dZ3 = pred3 - classes
    
    dw3 = 1/rows * np.dot(dZ3.T,pred2.T)
    db3 = 1/rows * np.sum(np.sum(dZ3))
    dZ2 = np.dot(W3.T,dZ3.T) * ReLU_deriv(unac_2)
    dw2 = 1/rows * np.dot(dZ2,pred1.T)
    db2 = 1/rows * np.sum(dZ2)
    dZ1 = np.dot(W2.T,dZ2)* ReLU_deriv(unac_1)
    dw1 = 1/rows*np.dot(dZ1,images.T)
    db1 = 1/rows*np.sum(dZ1)
    
    return dw1,db1,dw2,db2,dw3,db3,dZ3
    
    
```

```python
learning_rate = 0.1
for _ in range(1000):
    #this is pretty simple, it is calculating dot product each time loop goes off
    unactivated_output1 = forward_prop(weight_layer1,images,bias_layer1)
    neuron_output_layer1 = relu(unactivated_output1)
    unactivated_output2 = forward_prop(weight_layer2,neuron_output_layer1,bias_layer2)
    neuron_output_layer2 = relu(unactivated_output2)
    unactivated_output3 = forward_prop(weight_layer3,neuron_output_layer2,bias_layer3)
    neuron_output_layer3 = softmax(unactivated_output3)
    
    #here let's perform the back prop:::
    dw1,db1,dw2,db2,dw3,db3,dZ3=back_prop(images,classes,neuron_output_layer3,neuron_output_layer2,neuron_output_layer1,4000,weight_layer1,weight_layer2,weight_layer3,unactivated_output1,unactivated_output2,unactivated_output3)
    weight_layer1 -= learning_rate*dw1
    weight_layer2 -= learning_rate*dw2
    weight_layer3 -= learning_rate*dw3
    bias_layer1 -= learning_rate*db1
    bias_layer2 -= learning_rate*db2
    bias_layer3 -= learning_rate*db3
    if _%100== 0:
        print(dZ3)
        print("*"*80)
    warnings.filterwarnings("ignore", category=FutureWarning)
    
```

```python
def reverse_softmax(probabilities):
    # Add a small constant to avoid log(0)
    epsilon = 1e-12
    probabilities = np.clip(probabilities, epsilon, 1 - epsilon)
    
    # Compute logits
    logits = np.log(probabilities)
    logits -= np.max(logits, axis=-1, keepdims=True)  # Adjust for stability (optional)
    
    return logits
```

```python
def get_accuracy(predictions, Y):
    no = np.sum(predictions != Y)
    yes = np.sum(predictions == Y)
    return yes/(yes + no)
    
```
